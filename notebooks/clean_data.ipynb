## 3. Handling Missing Values

In this section, we will:

- Identify columns containing missing values
- Analyze the missing values
- Decide how each missing value should be handled
- Apply the appropriate cleaning strategy
- Verify the results

```python
df["Age"].median()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df["Age"].isna().sum()
```

```python
df["Salary"].median()
```

```python
df["Salary"] = df["Salary"].fillna(df["Salary"].median())
```

```python
df["Salary"].isna().sum()
```

```python
df.isna().sum()
```

## 4. Data Type Conversion

In this section, we will review the data types of the columns and convert columns whose current types do not match their intended meaning.

In particular, we will:

- Convert `Join_Date` from text to a datetime type
- Review the data type of `Phone`
- Ensure numeric columns use appropriate numeric types
- Verify the resulting data types

```python
df["Join_Date"] = pd.to_datetime(df["Join_Date"], format="%m/%d/%Y", errors="coerce")
```

```python
df["Join_Date"].info()
```

```python
df["Join_Date"].head(10)
```

```python
df["Join_Date"].isna().sum()
```

```python
df["Phone"].min()
```

```python
df["Phone"].max()
```

```python
df["Phone"] = df["Phone"].astype(str)
```

```python
df["Phone"].info()
```

```python
df["Phone"] = df["Phone"].str.strip("-")
```

```python
df["Phone"].str.len().value_counts()
```

```python
df.loc[df["Phone"].str.len() < 10, "Phone"].to_list()
```

```python
df.loc[df["Phone"].str.len() == 10, "Phone"].str[0].value_counts()
```

```python
df["Phone"] = df["Phone"].str.zfill(10)
```

```python
df["Phone"].str.len().value_counts()
```

```python
df["Phone"].head(10)
```

```python
df["Department_Region"].str.split("-", expand=True).head(10)
```

```python
df[["Department", "Region"]] = df["Department_Region"].str.split("-", expand=True)
```

```python
df["Age"] = df["Age"].astype(int)
df.head()
```

```python
columns_order = [
    "Employee_ID", "First_Name", "Last_Name", "Age",
    "Department", "Region", "Status", "Join_Date",
    "Salary", "Email", "Phone", "Performance_Score", "Remote_Work"
]
df = df[columns_order]
```
